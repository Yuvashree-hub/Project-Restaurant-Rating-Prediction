```python
# ============================================================
# RESTAURANT RATING PREDICTION USING MACHINE LEARNING
# ============================================================
# Dataset: Zomato Bangalore Restaurants
# Algorithm: Random Forest Regressor
# Target: Restaurant Rating
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv("zomato.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())


# ============================================================
# 3. BASIC DATASET INFORMATION
# ============================================================

print("\n================ DATASET INFORMATION ================")

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
display(df.isnull().sum())


# ============================================================
# 4. CLEAN RATING COLUMN
# ============================================================

# Convert values such as:
# 4.1/5 -> 4.1
# NEW -> NaN
# - -> NaN

df['rate'] = df['rate'].astype(str).str.split('/').str[0]

df['rate'] = pd.to_numeric(
    df['rate'],
    errors='coerce'
)

# Remove rows where rating is missing
df = df.dropna(subset=['rate'])

print("\nRating column cleaned successfully!")

print("\nRating Statistics:")
display(df['rate'].describe())


# ============================================================
# 5. CLEAN COST COLUMN
# ============================================================

df['approx_cost(for two people)'] = (
    df['approx_cost(for two people)']
    .astype(str)
    .str.replace(',', '', regex=False)
)

df['approx_cost(for two people)'] = pd.to_numeric(
    df['approx_cost(for two people)'],
    errors='coerce'
)

print("\nCost column cleaned successfully!")


# ============================================================
# 6. REMOVE DUPLICATE RESTAURANTS
# ============================================================

df = df.drop_duplicates()

print("\nDataset shape after removing duplicates:")
print(df.shape)


# ============================================================
# 7. EXPLORATORY DATA ANALYSIS
# ============================================================

print("\n================ EXPLORATORY DATA ANALYSIS ================")


# ------------------------------------------------------------
# 7.1 Rating Distribution
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

sns.histplot(
    df['rate'],
    bins=20,
    kde=True
)

plt.xlabel("Restaurant Rating")
plt.ylabel("Number of Restaurants")
plt.title("Distribution of Restaurant Ratings")

plt.show()


# ------------------------------------------------------------
# 7.2 Rating Count
# ------------------------------------------------------------

rating_counts = df['rate'].round(1).value_counts().sort_index()

plt.figure(figsize=(10, 5))

plt.bar(
    rating_counts.index.astype(str),
    rating_counts.values
)

plt.xlabel("Rating")
plt.ylabel("Number of Restaurants")
plt.title("Number of Restaurants by Rating")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# 7.3 Votes vs Rating
# ------------------------------------------------------------

sample_df = df.sample(
    min(5000, len(df)),
    random_state=42
)

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=sample_df,
    x='votes',
    y='rate',
    alpha=0.5
)

plt.xlabel("Number of Votes")
plt.ylabel("Restaurant Rating")
plt.title("Votes vs Restaurant Rating")

plt.show()


# ------------------------------------------------------------
# 7.4 Cost vs Rating
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=sample_df,
    x='approx_cost(for two people)',
    y='rate',
    alpha=0.5
)

plt.xlabel("Approximate Cost for Two People")
plt.ylabel("Restaurant Rating")
plt.title("Cost vs Restaurant Rating")

plt.show()


# ============================================================
# 8. TOP 10 RESTAURANTS BY ACTUAL RATING
# ============================================================

top_actual = (
    df[
        ['name', 'rate', 'votes']
    ]
    .sort_values(
        by=['rate', 'votes'],
        ascending=[False, False]
    )
    .head(10)
)

print("\n================ TOP 10 RESTAURANTS ================")

display(top_actual)


# ------------------------------------------------------------
# TOP RESTAURANT VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_actual,
    x='rate',
    y='name'
)

plt.xlabel("Restaurant Rating")
plt.ylabel("Restaurant Name")
plt.title("Top 10 Restaurants by Actual Rating")

plt.xlim(0, 5)

plt.show()


# ============================================================
# 9. TOP LOCATIONS BY AVERAGE RATING
# ============================================================

location_rating = (
    df.groupby('location')['rate']
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

print("\n================ TOP LOCATIONS ================")

display(location_rating)


# ------------------------------------------------------------
# LOCATION VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

sns.barplot(
    data=location_rating,
    x='rate',
    y='location'
)

plt.xlabel("Average Rating")
plt.ylabel("Location")
plt.title("Top 10 Locations by Average Restaurant Rating")

plt.xlim(0, 5)

plt.show()


# ============================================================
# 10. SELECT FEATURES FOR MACHINE LEARNING
# ============================================================

features = [
    'online_order',
    'book_table',
    'votes',
    'location',
    'rest_type',
    'cuisines',
    'approx_cost(for two people)'
]

target = 'rate'

X = df[features]
y = df[target]

print("\n================ FEATURES ================")

print("Input Features:")
for feature in features:
    print("-", feature)

print("\nTarget:")
print("-", target)


# ============================================================
# 11. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\n================ TRAIN TEST SPLIT ================")

print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])


# ============================================================
# 12. DEFINE NUMERICAL FEATURES
# ============================================================

numerical_features = [
    'votes',
    'approx_cost(for two people)'
]


# ============================================================
# 13. DEFINE CATEGORICAL FEATURES
# ============================================================

categorical_features = [
    'online_order',
    'book_table',
    'location',
    'rest_type',
    'cuisines'
]


# ============================================================
# 14. DATA PREPROCESSING
# ============================================================

numeric_transformer = Pipeline(
    steps=[
        (
            'imputer',
            SimpleImputer(strategy='median')
        )
    ]
)


categorical_transformer = Pipeline(
    steps=[
        (
            'imputer',
            SimpleImputer(strategy='most_frequent')
        ),
        (
            'onehot',
            OneHotEncoder(
                handle_unknown='ignore'
            )
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            numeric_transformer,
            numerical_features
        ),
        (
            'cat',
            categorical_transformer,
            categorical_features
        )
    ]
)


print("\nPreprocessing pipeline created successfully!")


# ============================================================
# 15. CREATE RANDOM FOREST MODEL
# ============================================================

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)


# ============================================================
# 16. CREATE COMPLETE MACHINE LEARNING PIPELINE
# ============================================================

pipeline = Pipeline(
    steps=[
        (
            'preprocessor',
            preprocessor
        ),
        (
            'model',
            model
        )
    ]
)


# ============================================================
# 17. TRAIN THE MODEL
# ============================================================

print("\n================ MODEL TRAINING ================")

print("Training Random Forest model...")

pipeline.fit(
    X_train,
    y_train
)

print("Model training completed successfully!")


# ============================================================
# 18. MAKE PREDICTIONS
# ============================================================

y_pred = pipeline.predict(X_test)

print("\n================ PREDICTIONS ================")

print("First 10 Predictions:")

print(
    np.round(
        y_pred[:10],
        2
    )
)


# ============================================================
# 19. ACTUAL VS PREDICTED RATINGS
# ============================================================

comparison = pd.DataFrame(
    {
        'Actual Rating': y_test.values,
        'Predicted Rating': np.round(
            y_pred,
            2
        )
    }
)

print("\nActual vs Predicted Ratings:")

display(
    comparison.head(10)
)


# ============================================================
# 20. MODEL EVALUATION
# ============================================================

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)


print("\n================ MODEL PERFORMANCE ================")

print(
    "Mean Absolute Error (MAE):",
    round(mae, 4)
)

print(
    "Mean Squared Error (MSE):",
    round(mse, 4)
)

print(
    "Root Mean Squared Error (RMSE):",
    round(rmse, 4)
)

print(
    "R2 Score:",
    round(r2, 4)
)


# ============================================================
# 21. ACTUAL VS PREDICTED VISUALIZATION
# ============================================================

plt.figure(figsize=(9, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.5
)

plt.xlabel("Actual Rating")
plt.ylabel("Predicted Rating")
plt.title("Actual vs Predicted Restaurant Ratings")

plt.plot(
    [1, 5],
    [1, 5],
    linestyle='--'
)

plt.xlim(1, 5)
plt.ylim(1, 5)

plt.show()


# ============================================================
# 22. KEEP RESTAURANT NAMES WITH PREDICTIONS
# ============================================================

# X_test contains the original indexes from df.
# Use those indexes to retrieve restaurant names.

test_restaurants = df.loc[
    X_test.index,
    ['name', 'rate']
].copy()


# Add predictions
test_restaurants['Predicted Rating'] = y_pred


# Round predictions
test_restaurants['Predicted Rating'] = (
    test_restaurants['Predicted Rating']
    .round(2)
)


# Calculate prediction difference
test_restaurants['Difference'] = (
    test_restaurants['rate'] -
    test_restaurants['Predicted Rating']
).abs().round(2)


print("\n================ RESTAURANT PREDICTIONS ================")

display(
    test_restaurants.head(20)
)


# ============================================================
# 23. TOP 10 RESTAURANTS BY PREDICTED RATING
# ============================================================

top_predicted = (
    test_restaurants
    .sort_values(
        by='Predicted Rating',
        ascending=False
    )
    .head(10)
)

print("\n================ TOP 10 PREDICTED RESTAURANTS ================")

display(
    top_predicted[
        [
            'name',
            'rate',
            'Predicted Rating',
            'Difference'
        ]
    ]
)


# ============================================================
# 24. VISUALIZE TOP 10 PREDICTED RESTAURANTS
# ============================================================

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_predicted,
    x='Predicted Rating',
    y='name'
)

plt.xlabel("Predicted Rating")
plt.ylabel("Restaurant Name")
plt.title("Top 10 Restaurants by Predicted Rating")

plt.xlim(0, 5)

plt.show()


# ============================================================
# 25. ACTUAL VS PREDICTED FOR TOP 10 RESTAURANTS
# ============================================================

top_comparison = (
    test_restaurants
    .sort_values(
        by='Predicted Rating',
        ascending=False
    )
    .head(10)
)

x = np.arange(
    len(top_comparison)
)

width = 0.35


plt.figure(figsize=(14, 7))

plt.bar(
    x - width / 2,
    top_comparison['rate'],
    width,
    label='Actual Rating'
)

plt.bar(
    x + width / 2,
    top_comparison['Predicted Rating'],
    width,
    label='Predicted Rating'
)

plt.xticks(
    x,
    top_comparison['name'],
    rotation=45,
    ha='right'
)

plt.xlabel("Restaurant")
plt.ylabel("Rating")
plt.title("Actual vs Predicted Ratings of Top Restaurants")

plt.ylim(0, 5)

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 26. FIND A SPECIFIC RESTAURANT
# ============================================================

restaurant_name = "Jalsa"


restaurant_data = df[
    df['name'].str.lower()
    == restaurant_name.lower()
]


if len(restaurant_data) > 0:

    restaurant_row = restaurant_data.iloc[0]


    # Create input data
    new_restaurant = pd.DataFrame(
        {
            'online_order': [
                restaurant_row['online_order']
            ],

            'book_table': [
                restaurant_row['book_table']
            ],

            'votes': [
                restaurant_row['votes']
            ],

            'location': [
                restaurant_row['location']
            ],

            'rest_type': [
                restaurant_row['rest_type']
            ],

            'cuisines': [
                restaurant_row['cuisines']
            ],

            'approx_cost(for two people)': [
                restaurant_row[
                    'approx_cost(for two people)'
                ]
            ]
        }
    )


    # Predict rating
    predicted_rating = pipeline.predict(
        new_restaurant
    )[0]


    print("\n============================================")
    print("       RESTAURANT RATING PREDICTION")
    print("============================================")

    print(
        "Restaurant Name :",
        restaurant_row['name']
    )

    print(
        "Actual Rating   :",
        restaurant_row['rate']
    )

    print(
        "Predicted Rating:",
        round(predicted_rating, 2)
    )

    print("============================================")


else:

    print(
        "Restaurant '",
        restaurant_name,
        "' was not found."
    )


# ============================================================
# 27. CREATE FINAL RESULTS TABLE
# ============================================================

final_results = test_restaurants[
    [
        'name',
        'rate',
        'Predicted Rating',
        'Difference'
    ]
].copy()


final_results.columns = [
    'Restaurant Name',
    'Actual Rating',
    'Predicted Rating',
    'Prediction Error'
]


print("\n================ FINAL RESTAURANT RESULTS ================")

display(
    final_results
    .sort_values(
        by='Predicted Rating',
        ascending=False
    )
    .head(20)
)


# ============================================================
# 28. FINAL PROJECT SUMMARY
# ============================================================

print("\n")
print("============================================================")
print("        RESTAURANT RATING PREDICTION PROJECT")
print("============================================================")

print(
    "Algorithm Used       : Random Forest Regressor"
)

print(
    "Number of Features   :",
    len(features)
)

print(
    "Training Samples     :",
    len(X_train)
)

print(
    "Testing Samples      :",
    len(X_test)
)

print(
    "MAE                  :",
    round(mae, 4)
)

print(
    "RMSE                 :",
    round(rmse, 4)
)

print(
    "R2 Score              :",
    round(r2, 4)
)

print("============================================================")
print("Restaurant Rating Prediction Completed Successfully!")
print("============================================================")
```
